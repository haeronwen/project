1.3 README.md  
Prague Student Cost of Living Simulator   
An interactive simulator that helps students estimate their monthly cost of living 
in Prague by selecting their preferred accommodation, groceries, and meal options.



Data Sources
- **Apartments**: Scraped from Bezrealitky.cz (1,363 listings)
- **Groceries**: Prices from Lidl.cz and Rohlik.cz
- **Dorms**: Charles University accommodation system (rehos.cuni.cz)
- **Canteen**: Charles University canteen API (kamweb.ruk.cuni.cz)
- **Other costs**: Czech Statistical Office data
