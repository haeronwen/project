```python
import pandas as pd
import numpy as np
from pathlib import Path 
import os 
import matplotlib.pyplot as plt
```

```python
path = os.path.abspath(os.path.join(os.getcwd(), '..'))
path = os.path.join(path, "data", "raw", "dorm_prices.csv")

dorms = pd.read_csv(path)
```

```python
dorms["beds"] = dorms["room_type"].str.split("-").str[0].astype(int)
```

```python
dorms["facilities"] = dorms["room_type"].str.contains("se soc. zař.", case=False, na=False).map({True: "Yes", False: "No"})
```

```python
dorms = dorms.rename(columns={"price": "monthly_cost"})
dorms
```

```python
dorm_stats = dorms.groupby(
    ["beds", "facilities"]
).agg(
    avg_cost=("monthly_cost", "mean"),
    count=("monthly_cost", "size")
).reset_index()
```

```python
dorm_stats
```

```python
os.chdir("..")
dorm_stats.to_csv("data/clean/dorm_stats.csv", index=False)
```
