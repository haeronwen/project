```python
pathl = os.path.abspath(os.path.join(os.getcwd(), '..'))
pathl = os.path.join(pathl, "data", "raw", "lidl_prices.csv")

lidl = pd.read_csv(pathl)
```

```python
pathr = os.path.abspath(os.path.join(os.getcwd(), '..'))
pathr = os.path.join(pathr, "data", "raw", "rohlik_prices.csv")

rohlik = pd.read_csv(pathr)
```

```python
groceries = pd.concat([rohlik, lidl], ignore_index=True)
```

```python
groceries = groceries.drop(columns=["item"])
```

```python
items = [
    "bread",
    "milk",
    "egg_salad",
    "butter",
    "chicken",
    "pasta",
    "rice",
    "chopped_tomatoes",
    "onion",
    "frozen_vegetables",
    "yoghurt",
    "coffee",
    "banana",
    "pastry",
    "flour",
    "pasta",
    "pastry",
    "chicken",
    "pasta",
    "rice",
    "pastry",
    "onion",
    "avocado",
    "yoghurt",
    "pasta",
    "fruit_puree"
]
```

```python
groceries["item"] = items
```

```python
grocery_stats = groceries.groupby("item").agg(
    avg_price=("price", "mean"),
    count=("price", "count")
).reset_index()
```

```python
basket={
    "bread":4,
    "milk":4,
    "egg_salad":2,
    "butter":1,
    "chicken":4,
    "pasta":4,
    "rice":2,
    "chopped_tomatoes":4,
    "onion":8,
    "frozen_vegetables":3,
    "yoghurt":10,
    "coffee":1,
    "banana":12,
    "flour":1
}
```

```python
def build_cheapest_basket(df,basket):
    #pick cheapest product per item category
    cheapest=(
        df[df["item"].isin(basket.keys())]
        .groupby("item")["price"]
        .min()
        .reset_index()
        .rename(columns={"price":"unit_price"}))
    #monthly quantity
    cheapest["qty_per_month"]=cheapest["item"].map(basket)
    #monthly cost per item 
    cheapest["monthly_cost"]=cheapest["unit_price"]*cheapest["qty_per_month"]

    return cheapest

result=build_cheapest_basket(groceries,basket)

print(result)
print(f"Total monthly cost: {result["monthly_cost"].sum():.2f} CZK")
```

```python
monthly_groceries=result["monthly_cost"].sum()
monthly_groceries
```
